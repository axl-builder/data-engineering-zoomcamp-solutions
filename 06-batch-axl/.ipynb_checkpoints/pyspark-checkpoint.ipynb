```python
import pyspark
from pyspark.sql import SparkSession
```

```python

spark = SparkSession.builder \
    .master("local[*]") \
    .appName('test') \
    .config("spark.driver.memory", "8g") \
    .config("spark.executor.memory", "4g") \
    .getOrCreate()
```

```python
!wget https://d37ci6vzurychx.cloudfront.net/trip-data/fhvhv_tripdata_2021-01.parquet
```

```python
df = spark.read.parquet('fhvhv_tripdata_2021-01.parquet')
```

```python
df.schema
```

```python
df_sample = df.limit(1000)
```

```python
df_sample.write.mode("overwrite").option("header", "true").csv('head_sample_csv')
```

```python
df.show(10)
```

```python
df.verticalDisplay = True # Solo en algunos entornos de notebook
df.show(1, vertical=True)
```
